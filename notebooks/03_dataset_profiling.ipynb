### Why use `UNKNOWN` instead of `mode()`?

Two missing values were found in the dataset:

- `object_type` → 2 missing values
- `country` → 2 missing values

Although these missing values can be filled using the **mode** (most frequent value), doing so may introduce incorrect information. For example, filling a missing `country` with `USA` or a missing `object_type` with `PAYLOAD` assumes the missing value belongs to the most common category, which cannot be verified.

Instead, these missing values will be replaced with **`UNKNOWN`** during the data cleaning phase. This approach:

- Preserves all records in the dataset.
- Avoids introducing incorrect or misleading information.
- Clearly indicates that the original value was unavailable.
- Maintains better data quality and transparency.
