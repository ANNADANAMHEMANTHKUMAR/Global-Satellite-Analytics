### Task 2: Distribution Analysis

- Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(df["mean_motion"], bins=30)

plt.title("Mean Motion Distribution")
plt.xlabel("Mean Motion")
plt.ylabel("Frequency")
plt.show()
```

- Skewness

```python
df["mean_motion"].skew()
```

- Kurtosis

```python
df["mean_motion"].kurt()
```

- Country Distribution

```python
country = df["country"].value_counts().head(10)

country.plot(kind="bar")

plt.title("Top Countries")
plt.xlabel("Country")
plt.ylabel("Count")
plt.show()
```

# Sampling Caveats

- The dataset represents a snapshot of currently available satellite records.
- It may not include every satellite launched worldwide.
- Country information depends on the available source data.
- Results are descriptive and may change when new satellite records are added.
- Insights should not be generalized beyond this dataset without additional validation.
