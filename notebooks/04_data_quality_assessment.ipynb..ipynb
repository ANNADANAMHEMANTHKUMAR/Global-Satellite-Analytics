## Objectives

- Detect data type issues.
- Identify invalid or out-of-range values.
- Check category consistency.
- Document all findings before data cleaning.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/raw/current_catalog.csv")
df1 = df.copy()
```

## 3. Category Consistency

Categorical columns are checked for inconsistent or unexpected category values.

## Findings

### Data Type Assessment

- No major data type issues were observed.
- Numeric columns were stored correctly.
- `launch_year_estimate` is stored as an integer (year), which is appropriate.
- Two categorical columns (`object_type` and `country`) contain two missing values each.

### Invalid Values

- No negative values were detected in `mean_motion`.
- Country values were reviewed for consistency.
- No invalid values were detected in the inspected numeric columns.

### Category Consistency

- Categories in `object_type` and `altitude_category` were consistent.
- `eccentricity` values were within the expected range (0–1).

Overall, the dataset is suitable for the data cleaning phase.
