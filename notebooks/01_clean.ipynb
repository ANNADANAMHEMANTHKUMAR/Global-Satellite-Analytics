## Purpose
This notebook loads the raw satellite dataset, performs initial data cleaning, handles missing values, standardizes data, and prepares the dataset for analysis.

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/raw/current_catalog.csv")
```

```python
df.head()
```

```python
df1 = df.copy()

print("Dataset copied successfully.")
```

```python
df1[["object_type", "country"]].isnull().sum()
```

```python
# Fill missing values
df1["object_type"] = df1["object_type"].fillna("UNKNOWN")
df1["country"] = df1["country"].fillna("UNKNOWN")

print("Cleaning completed successfully.")
```

```python
df1[["object_type", "country"]].isnull().sum()
```

```python
print("Original Rows:", len(df))
print("Cleaned Rows :", len(df1))

assert len(df) == len(df1), "Row count mismatch!"

print("✅ Row count validation passed.")
```

```python
assert df1["object_type"].isnull().sum() == 0
assert df1["country"].isnull().sum() == 0

print("✅ Null count validation passed.")
```

```python
df1.to_csv("../data/processed/current_catalog_cleaned.csv", index=False)

print("✅ Cleaned dataset saved successfully.")
```
